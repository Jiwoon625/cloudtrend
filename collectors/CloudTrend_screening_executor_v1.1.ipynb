# CloudTrend 스크리닝 실행기 v1.1

Drive에 이미 수집된 한국 주식·국내 ETF 자료로 한 번 계산하고 홈페이지에 결과를 게시합니다.
기존 v1·v6.1 파일은 바꾸지 않습니다.

- **입력 CSV와 과거 원천은 Supabase에 업로드하지 않습니다.**
- 홈페이지에는 계산 결과·대시보드 요약·출처와 완료 영수증만 저장합니다.
- 이전에 이 실행기로 사용한 Drive 추가분도 이어 읽어 날짜 사이 이력을 보존합니다.
- PRELIMINARY는 잠정으로 표시합니다. 이 실행기는 확정 신호 이력·원장·Shadow를 재작성하지 않습니다.
- 기본 과거 원천은 기존 Supabase 활성 자료를 읽습니다. 과거 원천 전체의 Drive 전환은 별도 v7 작업입니다.
- 최근 120일 차트도 결과로 준비할 수 있습니다. 홈페이지 표를 먼저 게시한 뒤 차트를 준비합니다. 전체 기간 차트는 별도 지원이 필요합니다.
- 결과는 원래 Drive run 폴더의 `screening-results-...zip`에도 보관합니다.

Colab 비밀의 `GITHUB_TOKEN`에 Actions·Secrets 쓰기 권한을 허용한 뒤 아래 두 셀을 실행하세요.
실행기만 실행하며 자료수집은 반복하지 않습니다. 값은 출력·저장되지 않습니다.

일반 실행은 **한 번 계산 → 홈페이지 게시 → 저장 결과 읽기 검증** 순서입니다.
추가 비교용 엔진 실행이나 원천 재등록을 하지 않습니다. 게시가 끝나면 홈페이지 링크를 표시합니다.

[CloudTrend 홈페이지](https://cloudtrend.vercel.app)


```python
# 수집이 완료된 Drive run을 선택하세요.
DRIVE_RUN_FOLDER_ID = "1dmPKBphuAm-6xXB-FCujuCIsgz5FtM4Z" #@param {type:"string"}
COLLECTOR_RUN_ID = "288f696eb14c4ef48f267f3fcfb61151" #@param {type:"string"}
WAIT_FOR_COMPLETION = True #@param {type:"boolean"}
PREPARE_RECENT_CHARTS = True #@param {type:"boolean"}
```

```python
# 스크리닝 후 결과만 홈페이지에 게시합니다.
import pathlib, subprocess, sys, urllib.request, hashlib
subprocess.run([sys.executable, "-m", "pip", "-q", "install", "PyNaCl==1.5.0"], check=True)
launcher = pathlib.Path("/content/cloudtrend_launch_drive_screening_v1_1.py")
url = "https://raw.githubusercontent.com/Jiwoon625/cloudtrend/main/scripts/launch-drive-screening-v1_1.py"
raw = urllib.request.urlopen(url, timeout=60).read()
if hashlib.sha256(raw).hexdigest() != "0995bb3d194ed0d0094bb22105c256287b4fe3062238949eaf7c68ca191a6e5d":
    raise RuntimeError("실행기 버전이 달라졌습니다. Drive의 최신 v1.1 파일을 다시 열어 주세요.")
launcher.write_bytes(raw)
namespace = {"__name__": "cloudtrend_launcher_v1_1"}
exec(compile(raw, str(launcher), "exec"), namespace)
namespace["launch"](DRIVE_RUN_FOLDER_ID, COLLECTOR_RUN_ID, "publish", wait=WAIT_FOR_COMPLETION, prepare_charts=PREPARE_RECENT_CHARTS)
```
