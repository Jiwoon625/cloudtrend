| 실행 방식 | 입력 처리 | 결과 |
| --- | --- | --- |
| `drive` (기본) | Drive의 저장 CSV를 직접 읽음 | 같은 Drive run 폴더에 결과 ZIP 저장. 홈페이지는 갱신하지 않음 |
| `supabase` | 기존 업로드·등록 절차 수행 | 홈페이지 갱신 및 Drive 결과 ZIP 저장 |

오늘 입력은 당일 추가분이므로, 두 방식 모두 기존 Supabase 과거 자료를 읽어 병합합니다.
Drive 방식은 Supabase 원본·등록부·홈페이지·원장에 쓰지 않습니다. 기존 수집기도 그대로 사용할 수 있습니다.

Fine-grained token은 해당 저장소의 **Actions: Read and write, Secrets: Read and write, Metadata: Read** 권한이 필요합니다.
Google 로그인은 실행할 때마다 진행합니다. 인증값은 출력하거나 결과 파일에 넣지 않고 GitHub Secret으로 암호화 전송합니다.

```python
# 실행기 코드를 준비하고 스크리닝 시작 (수집 없음)
import pathlib, subprocess, sys, urllib.request
subprocess.run([sys.executable, "-m", "pip", "-q", "install", "PyNaCl==1.5.0"], check=True)
launcher = pathlib.Path("/content/cloudtrend_launch_drive_screening.py")
url = "https://raw.githubusercontent.com/Jiwoon625/cloudtrend/main/scripts/launch-drive-screening.py"
launcher.write_bytes(urllib.request.urlopen(url, timeout=60).read())
namespace = {"__name__": "cloudtrend_launcher"}
exec(compile(launcher.read_bytes(), str(launcher), "exec"), namespace)
namespace["launch"](DRIVE_RUN_FOLDER_ID, COLLECTOR_RUN_ID, SOURCE_STORAGE)
```
